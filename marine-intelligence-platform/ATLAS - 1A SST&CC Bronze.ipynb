```sql
-- What is CC?
-- Chlorophyll Concentration — how much phytoplankton (tiny ocean plants)
-- is present. More chlorophyll = more food for filter feeding species.
-- Measured in milligrams per cubic metre (mg/m³).
--
-- Why exclude FullYear files?
-- Each folder contains annual average files (e.g. CC-FullYear2009.csv).
-- These are different data — yearly averages not monthly readings.
-- We keep monthly files only here.
--
-- Note: May 2024 does not exist from NOAA — known source gap.
```

```sql
-- CC 2009

CREATE OR REPLACE TABLE mcphersonsharyn_bronze.bronze_cc_2009 AS
SELECT
  `Latitude`            AS latitude_raw,
  ` Longitude`          AS longitude_raw,   -- Note - space before Longitude
  ` Data Value`         AS cc_raw,
  2009                  AS source_year,
  _metadata.file_name   AS source_table,
  current_timestamp()   AS ingested_timestamp
FROM read_files('/Volumes/workspace/raw/environmental_data/2009/',
  format => 'csv', header => true)
WHERE _metadata.file_name NOT LIKE '%FullYear%';
 
```

```sql
-- Quick check — should show 2009 with rows > 0

SELECT source_year, count(*) AS rows FROM mcphersonsharyn_bronze.bronze_cc_2009 GROUP BY source_year;
```

```sql
-- What is SST?
-- Sea Surface Temperature — how warm the ocean surface is in degrees
-- Celsius. Monthly readings from NOAA for our study region.
--
-- IMPORTANT — SST folder naming is inconsistent in your volume:
--   2009 = SST- 2009  (no space before the dash)
--   All others = SST - 2010, SST - 2012 etc. (space before the dash)
--   The exact paths below match what was confirmed by the LIST command.
--
-- SST 2022 was NOT found in the volume listing.
--   Check your volume before running the SST 2022 query.
--   If missing — upload the SST 2022 folder then run that query. 

--  SST 2022 was uploaded
```
