```python
%sql
--   ATLAS  - Practise Visualisation Queries
--   Three charts to run first in your ATLAS_Visualisations notebook

-- Chart 1. = Average sea surface temperature for each year in your study period.
-- Each ENSO phase (El Nino, La Nina, Neutral) appears as a separate line.

-- This is the first visual view of your key finding.
-- If La Nina years in 2020-2022 sit HIGHER on the chart than El Nino years in 2014-2015 — the long-term warming signal is overriding
-- Expected ENSO cooling effect in your study region.


-- What to look for:
-- La Nina lines should be BELOW El Nino lines in early years
-- La Nina lines should be ABOVE El Nino lines in recent years
-- That crossover is the ATLAS finding made visible
--
-- Databricks chart settings:
--   Chart type : Line
--   X axis     : year
--   Y axis     : avg_sst
--   Group by   : enso_phase
--   Label      : colour each phase differently
--                El Nino = warm colour
--                La Nina = cool colour
--                Neutral = grey
-- ─────────────────────────────────────────────────────────────────────────

SELECT
year,
round(avg(sst), 2) AS avg_sst,
enso_phase
FROM mcphersonsharyn_gold.gold_environment_monthly
WHERE enso_phase IS NOT NULL
GROUP BY year, enso_phase
ORDER BY year;


-- ███████████████████████████████████████████████████████████████████████████
-- ██                                                                       ██
-- ██   QUICK START CHART 2 — ENSO PHASE SST AND CC COMPARISON             ██
-- ██   Bar chart — run this second — your key finding in three bars       ██
-- ██                                                                       ██
-- ███████████████████████████████████████████████████████████████████████████

-- ─────────────────────────────────────────────────────────────────────────
-- What this shows:
-- Three bars side by side — one per ENSO phase — showing the overall
-- average SST and average chlorophyll for that phase across the full
-- study period 2009 to 2025.
--
-- Why this matters:
-- This is the simplest and most powerful version of your key finding.
-- In just three bars it shows:
--   La Nina average SST = 29.47 degrees C  (tallest bar — UNEXPECTED)
--   Neutral average SST = 29.31 degrees C
--   El Nino average SST = 28.80 degrees C  (shortest bar — UNEXPECTED)
--
-- And for chlorophyll — the reverse:
--   El Nino average CC = 0.426  (most food — UNEXPECTED)
--   La Nina average CC = 0.367  (least food — UNEXPECTED)
--
-- What to look for:
-- La Nina bar should be the TALLEST for SST
-- El Nino bar should be the TALLEST for CC
-- Both are the opposite of what climate science predicts
-- This is the chart to screenshot and share
--
-- Databricks chart settings:
--   Chart type : Bar
--   X axis     : enso_phase
--   Y axis     : avg_sst for first chart
--                avg_cc for second chart (run same query twice)
--   Sort       : ordered by avg_sst DESC so La Nina appears first
--
-- Note: years_in_phase column tells you how many years of data
-- contribute to each bar — useful for understanding the sample size
-- ─────────────────────────────────────────────────────────────────────────

SELECT
enso_phase,
round(avg(sst), 2) AS avg_sst,
round(avg(cc), 3) AS avg_cc,
count(DISTINCT year) AS years_in_phase
FROM mcphersonsharyn_gold.gold_environment_monthly
WHERE enso_phase IS NOT NULL
GROUP BY enso_phase
ORDER BY avg_sst DESC;


-- ███████████████████████████████████████████████████████████████████████████
-- ██                                                                       ██
-- ██   QUICK START CHART 3 — SPECIES PRESENCE BY YEAR                     ██
-- ██   Line chart — run this third                                        ██
-- ██                                                                       ██
-- ███████████████████████████████████████████████████████████████████████████

-- ─────────────────────────────────────────────────────────────────────────
-- What this shows:
-- The number of unique tagged animals recorded each year for each species.
-- One line per species — four lines total on the same chart.
--
-- Why this matters:
-- This tells you whether the number of animals observed in each study
-- area is changing over time. A declining line could indicate:
--   - Fewer animals using that habitat
--   - A response to changing environmental conditions
--   - The end of a tagging programme (important caveat — see below)
--
-- What to look for:
-- Whale shark — peaks at 14 animals in 2014 then drops to 6 in 2015
--               Does this coincide with worsening conditions?
-- Sperm whale — drops from 11 animals in 2011 to 6 in 2013
--               This 45 percent drop may relate to Deepwater Horizon
-- Green turtle — volatile pattern 2012 to 2019 with a spike to 16 in 2018
--               Not a simple decline — habitat appears viable throughout
-- Reef manta ray — Raja Ampat shows presence across 2014 to 2022
--                  Other localities disappear after 2015
--
-- Important caveat:
-- A decline in unique animals does not necessarily mean fewer animals
-- exist in the wild. It may reflect fewer animals being tagged in later
-- years of the study. Always check how many animals were tagged per year
-- before drawing population-level conclusions from this chart.
--
-- Databricks chart settings:
--   Chart type    : Line
--   X axis        : year
--   Y axis        : unique_animals
--   Group by      : species_common
--   Second metric : total_observations (add as a second line or tooltip)
-- ─────────────────────────────────────────────────────────────────────────

SELECT
year,
species_common,
count(DISTINCT animal_id) AS unique_animals,
count(*) AS total_observations
FROM mcphersonsharyn_silver.migration_silver
GROUP BY year, species_common
ORDER BY species_common, year;


-- ═══════════════════════════════════════════════════════════════════════════
--
--   SUGGESTED RUN ORDER FOR THESE THREE CHARTS
--   ────────────────────────────────────────────
--   1. Chart 2 first  — three bars, instant visual of your key finding
--   2. Chart 1 second — the full year by year picture with ENSO lines
--   3. Chart 3 third  — species presence, the conservation story
--
--   Once these three are working and looking good
--   move to Workbook 6 for the full set of 10 visualisation queries
--
-- ═══════════════════════════════════════════════════════════════════════════
```
