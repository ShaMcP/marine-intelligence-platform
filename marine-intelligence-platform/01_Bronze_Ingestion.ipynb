```sql
SELECT
  count(*)                                              AS total_rows,
  count(*) filter(WHERE latitude_raw  IS NULL)          AS null_lat,
  count(*) filter(WHERE longitude_raw IS NULL)          AS null_lon,
  count(*) filter(WHERE sst_raw       IS NULL)          AS null_sst,
  count(*) filter(WHERE latitude_raw  IS NOT NULL
    AND longitude_raw IS NOT NULL
    AND CAST(latitude_raw  AS DOUBLE) NOT BETWEEN -90  AND 90)
                                                        AS invalid_lat,
  count(*) filter(WHERE latitude_raw  IS NOT NULL
    AND longitude_raw IS NOT NULL
    AND CAST(longitude_raw AS DOUBLE) NOT BETWEEN -180 AND 180)
                                                        AS invalid_lon
FROM workspace.mcphersonsharyn_bronze.bronze_sst;
```
