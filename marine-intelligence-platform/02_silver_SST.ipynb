```sql
 create schema if not exists workspace.mcphersonsharyn_silver;
```

```sql
CREATE OR REPLACE TABLE workspace.mcphersonsharyn_silver.sea_surface_temperature_silver AS
SELECT
  CAST(`Latitude` AS DOUBLE)                          AS latitude,
  CAST(` Longitude` AS DOUBLE)                        AS longitude,
  ` Data Value`                                       AS sea_surface_temperature,
  '2025'                                              AS year,
  '01'                                                AS month,
  'monthly'                                           AS temporal_resolution,
  22                                                  AS lat_upper_bound,
  -23                                                 AS lat_lower_bound,
  126                                                 AS long_upper_bound,
  -73                                                 AS long_lower_bound,
  h3_longlatash3(
    CAST(` Longitude` AS DOUBLE),
    CAST(`Latitude` AS DOUBLE),
    5)                                                AS h3_res5,
  h3_longlatash3(
    CAST(` Longitude` AS DOUBLE),
    CAST(`Latitude` AS DOUBLE),
    6)                                                AS h3_res6,
  current_timestamp()                                 AS processed_timestamp
FROM
  workspace.mcphersonsharyn_bronze.bronze_sst_jan;
```
